```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x, y)
f1 = X**2 + Y**2
f2 = (X - 4)**2 + (Y - 4)**2
f3 = np.sin(X) + np.cos(Y)
f4 = np.abs(X - Y)
fig = plt.figure(figsize=(16,10))
ax1 = fig.add_subplot(221, projection='3d')
ax1.plot_surface(X, Y, f1, cmap='viridis',alpha = 0.7)
ax1.set_title('f1 = x² + y²')
ax2 = fig.add_subplot(222, projection='3d')
ax2.plot_surface(X, Y, f2, cmap='viridis',alpha = 0.7)
ax2.set_title('f2 = (x-4)² + (y-4)²')
ax3 = fig.add_subplot(223, projection='3d')
ax3.plot_surface(X, Y, f3, cmap='viridis',alpha = 0.7)
ax3.set_title('f3 = sin(x) + cos(y)')
ax4 = fig.add_subplot(224, projection='3d')
ax4.plot_surface(X, Y, f4, cmap='viridis',alpha = 0.7)
ax4.set_title('f4 = |x-y|')

plt.tight_layout()
plt.show()
```

```python

fig = plt.figure(figsize=(16,10))

ax1 = fig.add_subplot(221)

contour1 = ax1.contour(X, Y, f1,levels=20,colors='red')
contour2 = ax1.contour(X, Y, f2,levels=20,colors='blue',alpha=0.7,linewidths=1.5)

ax1.set_title('Curvas de nivel de $f_1$ y $f_2$')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_aspect('equal')
ax1.plot(0,0,'ro',label='Min f1')
ax1.plot(4,4,'bo',label='Min f2')

ax1.legend()

plt.show()
```
